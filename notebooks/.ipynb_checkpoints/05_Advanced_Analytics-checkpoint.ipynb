```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import percentileofscore

import warnings
warnings.filterwarnings('ignore')

plt.style.use('default')
```

```python
fact_performance = pd.read_csv("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/dashboard/tableau_data/fact_performance.csv")

fact_transactions = pd.read_csv("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/dashboard/tableau_data/fact_transactions.csv")

fact_portfolio = pd.read_csv("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/dashboard/tableau_data/fact_portfolio.csv")

fund_scorecard = pd.read_csv("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/dashboard/tableau_data/fund_scorecard.csv")

dim_fund = pd.read_csv("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/dashboard/tableau_data/dim_fund.csv")
```

```python
print(fact_performance.columns)
print(fact_transactions.columns)
print(fact_portfolio.columns)
print(fund_scorecard.columns)
print(dim_fund.columns)
```

```python
var_cvar_report = fact_performance.copy()

var_cvar_report["VaR_95"] = (
    var_cvar_report["return_3yr_pct"]
    - 1.65 * var_cvar_report["std_dev_ann_pct"]
)

var_cvar_report["CVaR_95"] = (
    var_cvar_report["VaR_95"] * 1.15
)

var_cvar_report = var_cvar_report[
    [
        "scheme_name",
        "fund_house",
        "category",
        "VaR_95",
        "CVaR_95"
    ]
]

var_cvar_report.to_csv(
    "var_cvar_report.csv",
    index=False
)

print(var_cvar_report.head())
```

```python
top5_sharpe = (
    fact_performance
    .sort_values(
        "sharpe_ratio",
        ascending=False
    )
    .head(5)
)

plt.figure(figsize=(10,5))

plt.bar(
    top5_sharpe["scheme_name"],
    top5_sharpe["sharpe_ratio"]
)

plt.xticks(rotation=45)
plt.title("Top 5 Funds by Sharpe Ratio")
plt.ylabel("Sharpe Ratio")

plt.tight_layout()

plt.savefig(
    "rolling_sharpe_chart.png"
)

plt.show()
```

```python
fact_transactions["transaction_date"] = pd.to_datetime(
    fact_transactions["transaction_date"]
)
```

```python
first_txn = (
    fact_transactions
    .groupby("investor_id")
    ["transaction_date"]
    .min()
    .reset_index()
)

first_txn["cohort_year"] = (
    first_txn["transaction_date"]
    .dt.year
)
```

```python
cohort_df = fact_transactions.merge(
    first_txn[
        [
            "investor_id",
            "cohort_year"
        ]
    ],
    on="investor_id"
)
```

```python
cohort_summary = (
    cohort_df
    .groupby("cohort_year")
    .agg(
        avg_amount=("amount_inr","mean"),
        total_invested=("amount_inr","sum")
    )
)

print(cohort_summary)
```

```python
top_fund = (
    cohort_df
    .groupby(
        [
            "cohort_year",
            "amfi_code"
        ]
    )
    .size()
    .reset_index(name="count")
)

top_fund = top_fund.sort_values(
    ["cohort_year","count"],
    ascending=[True,False]
)

top_fund = top_fund.groupby(
    "cohort_year"
).first()

print(top_fund)
```

```python
sip = fact_transactions[
    fact_transactions["transaction_type"]
    .str.upper()
    ==
    "SIP"
].copy()
```

```python
sip_count = (
    sip
    .groupby("investor_id")
    .size()
)

eligible = sip_count[
    sip_count >= 6
].index

sip = sip[
    sip["investor_id"]
    .isin(eligible)
]
```

```python
sip = sip.sort_values(
    [
        "investor_id",
        "transaction_date"
    ]
)

sip["gap_days"] = (
    sip.groupby("investor_id")
    ["transaction_date"]
    .diff()
    .dt.days
)
```

```python
gap_report = (
    sip.groupby("investor_id")
    ["gap_days"]
    .mean()
    .reset_index()
)

gap_report.rename(
    columns={
        "gap_days":"avg_gap_days"
    },
    inplace=True
)
```

```python
gap_report["status"] = np.where(
    gap_report["avg_gap_days"] > 35,
    "At-Risk",
    "Healthy"
)

print(gap_report.head())
```

```python
def recommend_funds(risk_level):

    rec = (
        fact_performance[
            fact_performance["risk_grade"]
            ==
            risk_level
        ]
        .sort_values(
            "sharpe_ratio",
            ascending=False
        )
        .head(3)
    )

    return rec[
        [
            "scheme_name",
            "fund_house",
            "sharpe_ratio",
            "return_3yr_pct"
        ]
    ]
```

```python
print(recommend_funds("High"))
```

```python
hhi = (
    fact_portfolio
    .groupby("amfi_code")
    .apply(
        lambda x:
        (
            (x["weight_pct"]/100)**2
        ).sum()
    )
    .reset_index(name="HHI")
)
```

```python
hhi = hhi.merge(
    dim_fund[
        [
            "amfi_code",
            "scheme_name"
        ]
    ],
    on="amfi_code"
)
```

```python
hhi = hhi.sort_values(
    "HHI",
    ascending=False
)

print(
    hhi[
        [
            "scheme_name",
            "HHI"
        ]
    ]
    .head(10)
)
```
