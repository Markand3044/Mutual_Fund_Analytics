```python
import pandas as pd
import os
```

```python
export_path = r"C:\Users\Admin\Desktop\Mutual_Fund_Analysis\dashboard\tableau_data"

os.makedirs(export_path, exist_ok=True)

print("Folder Ready")
```

```python
from sqlalchemy import create_engine

server = "laptop-t5cbj27g\SQLEXPRESS"
database = "Bluestock_MF"

engine = create_engine(
    f"mssql+pyodbc://@{server}/{database}"
    "?driver=ODBC+Driver+17+for+SQL+Server"
    "&trusted_connection=yes"
)

print("Connected Successfully")
```

```python
tables = [
    "dim_fund",
    "fact_aum",
    "fact_sip_industry",
    "fact_category_inflows",
    "fact_folio_count",
    "fact_performance",
    "fact_transactions",
    "fact_portfolio",
    "fact_benchmark"
]

for table in tables:

    df = pd.read_sql(
        f"SELECT * FROM {table}",
        engine
    )

    file_path = os.path.join(
        export_path,
        f"{table}.csv"
    )

    df.to_csv(
        file_path,
        index=False
    )

    print(f"{table} exported")
```
