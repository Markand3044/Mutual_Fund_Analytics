```python
import pandas as pd
from pathlib import Path

raw_path = Path("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/data/raw")

csv_files = list(raw_path.glob("*.csv"))

print(f"Found {len(csv_files)} CSV files")

anomalies = []

for file in csv_files:

    print("\n" + "="*80)
    print(f"FILE: {file.name}")

    df = pd.read_csv(file)

    # shape
    print("\nShape:")
    print(df.shape)

    # dtypes
    print("\nData Types:")
    print(df.dtypes)

    # head
    print("\nHead:")
    print(df.head())

    # anomalies
    if df.isnull().sum().sum() > 0:
        anomalies.append(f"{file.name} contains missing values")

    if df.duplicated().sum() > 0:
        anomalies.append(f"{file.name} contains duplicate rows")

print("\n" + "="*80)
print("ANOMALIES")

for a in anomalies:
    print(a)
```
