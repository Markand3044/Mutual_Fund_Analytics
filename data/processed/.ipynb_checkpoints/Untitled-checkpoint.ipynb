```python
import pandas as pd

nav = pd.read_csv("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/data/raw/02_nav_history.csv")

# date
nav['date'] = pd.to_datetime(nav['date'])

# sort
nav = nav.sort_values(['amfi_code','date'])

# forward fill nav
nav['nav'] = nav.groupby('amfi_code')['nav'].ffill()

# remove duplicates
nav = nav.drop_duplicates()

# nav > 0
nav = nav[nav['nav'] > 0]

nav.to_csv(
    "C:/Users/Admin/Desktop/Mutual_Fund_Analysis/data/processed/02_nav_history.csv",
    index=False
)
```
