```python
import pandas as pd
import numpy as np
```

```python
fact_performance = pd.read_csv("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/dashboard/tableau_data/fact_performance.csv")

fact_transactions = pd.read_csv("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/dashboard/tableau_data/fact_transactions.csv")

fact_portfolio = pd.read_csv("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/dashboard/tableau_data/fact_portfolio.csv")

fund_scorecard = pd.read_csv("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/dashboard/tableau_data/fund_scorecard.csv")

dim_fund = pd.read_csv("C:/Users/Admin/Desktop/Mutual_Fund_Analysis/dashboard/tableau_data/dim_fund.csv")
```

```python
def recommend_funds(risk_level):

    rec = (
        fact_performance[
            fact_performance["risk_grade"]
            ==
            risk_level
        ]
        .sort_values(
            "sharpe_ratio",
            ascending=False
        )
        .head(3)
    )

    return rec[
        [
            "scheme_name",
            "fund_house",
            "sharpe_ratio",
            "return_3yr_pct"
        ]
    ]
```
